```python
# Librerias
from qiskit import *
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```

```python
# Función para crear un oracle constante (responde siempre 0)
def oracle_constante(n):
    oracle = QuantumCircuit(n + 1)
    # No realiza operaciones, representa una funcion constante que devuelve 0
    return oracle

# Función para crear un oracle equilibrada
def oracle_equilibrada(n):
    oracle = QuantumCircuit(n + 1)
    # Aplicar CNOTs desde cada qubit de entrada al auxiliar
    for qubit in range(n):
        oracle.cx(qubit, n)
    return oracle

def deutsch_joza_nqubits(n, oracle):
    # Circuito con n qubits de entrada + 1 auxiliar
    qc = QuantumCircuit(n + 1, n)

    # Preparar estado inicial: superposición en los qubits de entrada
    for qubit in range(n):
        # Aplicar Compuerta Hadamard para Superposicion
        qc.h(qubit)
    # Estado auxiliar en |1>
    # Aplicar Compuerta NOT
    qc.x(n)
    # Aplicar segunda Compuerta Hadamard
    qc.h(n)

    # Agregar la oracle
    qc.compose(oracle, inplace=True)

    # Hadamard en los qubits de entrada
    for qubit in range(n):
        qc.h(qubit)

    # Medir los qubits de entrada
    qc.measure(range(n), range(n))
    # Dibujar Circuito
    qc.draw('mpl')

    # Ejecutar en simulador
    backend = Aer.get_backend('qasm_simulator')
    compilado = transpile(qc, backend)
    job = backend.run(compilado, shots=1024)
    # Obtener Resultados
    result = job.result()
    # Conteo
    counts = result.get_counts()
    # Plottear
    plot_histogram(counts)

    print(f"\nResultado para n={n} qubits {oracle}):")
    print(counts)

    # Interpretacion
    if '0' * n in counts:
        print("La función es constante.")
    else:
        print("La función es equilibrada.")

    # Frequencias
    for estado, frecuencia in counts.items():
        print(f"Estado: {estado} - Frecuencia: {frecuencia}")
```

```python
# Ejemplo de uso
n_qubits = 2

# Oracle constante
oracle_const = oracle_constante(n_qubits)
deutsch_joza_nqubits(n_qubits, oracle_const)

# Oracle equilibrada
oracle_bal = oracle_equilibrada(n_qubits)
deutsch_joza_nqubits(n_qubits, oracle_bal)
```
