```python
from qiskit import *
from qiskit_aer import Aer
# Visualizar los resultados
from qiskit.visualization import plot_histogram  #histograma
```

```python
qcirc = QuantumCircuit(2, 2)
qcirc.h(0)
qcirc.h(1)
qcirc.x(1)
qcirc.barrier()
```

```python
qcirc.measure(0, 0)
qcirc.measure(1, 1)
qcirc.draw('mpl')
```

```python
# Definir el simulador 'aer_simulator','statevector_simulator','qasm_simulator','unitary_simulator'
backend = Aer.get_backend("statevector_simulator")
```

```python
#Compilar y ejecutar el circuito en la computadora o simulador seleccionado el número de shots de 1024
compilado = transpile(qcirc, backend)
job = backend.run(compilado, shots=1024) #Ejecutar el circuito en la computadora o simulador seleccionado el número de shots dado
result = job.result() #Recupera los datos
outputstate = result.get_statevector(qcirc, decimals=3) #Visualizar el vector de estado (limita los resultados a 3 decimales)
print(outputstate)

```

```python
counts = job.result().get_counts()
plot_histogram(counts)
```
